Contour Plot- How Far Can A Store Deliver For AGiven Basket and Still Keep The Ten Minute Promise

```python
import numpy as np

#Developing Mesh Grid
# A tiny example first
#Step 1
distance = np.array([1, 2, 3])
basket   = np.array([5, 10])

D, B = np.meshgrid(distance, basket)

print(D)
print(B)

#Step 2-Create 40 possible delivery distance values and basket values
dist_values = np.linspace(0.3, 4, 40)
basket_values = np.linspace(1, 12, 40)

D, B = np.meshgrid(dist_values, basket_values)

print("Number of what-if cases:", D.size)

#Step 3: Linear Regression
from sklearn.linear_model import LinearRegression

formula = LinearRegression().fit(df[["Distance_KM", "Basket_Size"]], df["Delivery_Time_Min"])

print("Each extra km adds about", round(formula.coef_[0], 1), "minutes")
print("Each extra item adds about", round(formula.coef_[1], 2), "minutes")

#Predictions of Linear Model
# Fill every what-if case with the expected delivery time
cases = pd.DataFrame({"Distance_KM": D.ravel(), "Basket_Size": B.ravel()})
T = formula.predict(cases).reshape(D.shape)
#Step 5
plt.contourf(D, B, T, levels=10, cmap="Purples")
plt.colorbar(label="Expected delivery time (minutes)")

lines = plt.contour(D, B, T, levels=[8, 10, 12, 15], colors="red")
plt.clabel(lines, fmt="%d min")

plt.title("Distance and basket size that keep the 10-minute promise")
plt.xlabel("Distance from store (km)")
plt.ylabel("Basket size (items)")
plt.show()
```

Stastical modeling - liner equation and multiple regression

```python
import statsmodels.formula.api as smf

# Fit the simple linear regression model
m_social = smf.ols("Monthly_Sales_INR_Lakh ~ Social_Media_Spend_K", data=df).fit()

# Print the summary
print(m_social.summary())
print("R-squared:", round(m_social.rsquared, 2))
```

Interpretation of linear equation

a unit increase in social media spending is associated with 0.295 lakhs increase in monthly sales.
Social media spending is explained by 45% variation in monthly sales p Value/ significance level , social media spending as a significance of positive relationship with a monthly sales

Negative Relationship among variables in linear equation

```python
m_wait = smf.ols("Repeat_Customer_Pct ~ Avg_Wait_Time_Min", data=df).fit()
print(m_wait.summary().tables[1])
print("R-squared:", round(m_wait.rsquared, 2))
```

what are the other atttributes that can effect monthly sales
